## 5. Contact Inquiries Validation Pipeline
Simulating email validation, length thresholds, and spam filtering for visitor inquiries.

```python
import re

email_regex = r"^[^@]+@[^@]+\.[^@]+$"

test_inquiries = [
    {"sender": "Recruiter", "email": "talent@tech.corp", "message": "Hi Sai Rakesh, we were impressed by your Forge Terminal and would love to chat about a developer role!"},
    {"sender": "Collaborator", "email": "dev@open.source", "message": "Awesome work on CodePilot AI. Let us collaborate on the Gemini model!"},
    {"sender": "Invalid User", "email": "bad-email", "message": "Test"}
]

for inquiry in test_inquiries:
    is_valid_email = bool(re.match(email_regex, inquiry["email"]))
    if is_valid_email and len(inquiry["message"]) >= 10:
        print(f"Inquiry from {inquiry['sender']} ({inquiry['email']}): VALID (Length: {len(inquiry['message'])} chars)")
    else:
        print(f"Inquiry from {inquiry['sender']} ({inquiry['email']}): REJECTED (Invalid email format or too short)")

print("Validation pipeline test completed successfully.")
```
